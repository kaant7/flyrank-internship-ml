## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Apply the starter feature-prep rules from the lane guide
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

print("Total usable rows:", len(df))

# Number 1: enough scale to find real patterns, not a toy sample
declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"Share of pages currently declining: {declining_pct:.1f}%")

# Number 2: meaningful variation — not everything is the same, so there's something to explain
high_imp = df[df["impressions_90d"] >= 500]
low_imp = df[df["impressions_90d"] < 500]
print("Decline rate, high-impression pages:", (high_imp["trend_direction"] == "down").mean() * 100)
print("Decline rate, low-impression pages:", (low_imp["trend_direction"] == "down").mean() * 100)

# Number 3: a real, non-zero relationship between two observed signals
print("CTR - position correlation:", df["ctr"].corr(df["avg_position"]))
```

Why this looks worth it: With 30,000 usable pages, there's enough scale to detect real patterns rather than noise. Decline rates differ meaningfully between high- and low-impression pages (59.6% vs 47.5%) — an 12-point gap suggests visibility level isn't independent of trend direction, exactly the kind of signal-to-outcome relationship Lane 1 is built to investigate. The CTR-position correlation (-0.073) is weak but in the expected direction, worth digging into further with position-tier-adjusted comparisons rather than a raw correlation.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What my work will be able to say:

-Observed: Which signals (impressions, position, freshness, CTR, engagement) show measurable association with visibility and trend movement in this dataset.

-Directional: For example, pages with higher impressions show a higher decline rate than lower-impression pages (59.6% vs 47.5%), suggesting visibility level and trend direction are not independent.

-Decision-support: A ranked or flagged set of signals that a reviewer can use to prioritize which pages to look at first — narrowing down where limited review time goes, not making the call for them.

What my work will never say:

-Causal proof: A correlation between a signal and decline does not mean that signal caused the decline — I did not run an experiment, so I cannot claim cause and effect.

-"Predicting Google": I am not claiming to have found or reverse-engineered a Google ranking factor. These are observed correlations within FlyRank's own data, not evidence about how any search engine's algorithm works.

-Guaranteed outcomes: A page flagged as high-risk or high-opportunity is a candidate for review, not a guarantee that fixing it will recover traffic.
