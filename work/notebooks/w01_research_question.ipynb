## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Apply the starter feature-prep rules from the lane guide
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

print("Total usable rows:", len(df))

# Number 1: enough scale to find real patterns, not a toy sample
declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"Share of pages currently declining: {declining_pct:.1f}%")

# Number 2: meaningful variation — not everything is the same, so there's something to explain
high_imp = df[df["impressions_90d"] >= 500]
low_imp = df[df["impressions_90d"] < 500]
print("Decline rate, high-impression pages:", (high_imp["trend_direction"] == "down").mean() * 100)
print("Decline rate, low-impression pages:", (low_imp["trend_direction"] == "down").mean() * 100)

# Number 3: a real, non-zero relationship between two observed signals
print("CTR - position correlation:", df["ctr"].corr(df["avg_position"]))
```
